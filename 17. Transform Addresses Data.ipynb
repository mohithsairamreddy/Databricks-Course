## Transform Addresses Data
1. Create one record for each customer with 2 sets of address columns, 1 for shipping and 1 for billing address
2. Write transformed data to the Silver

```sql
SELECT * FROM gizmobox.bronze.v_addresses
```

```sql
%sql
SELECT
    customer_id,
    MAX(CASE WHEN LOWER(address_type) = 'shipping' THEN address_line_1 END) AS shipping_address_line_1,
    MAX(CASE WHEN LOWER(address_type) = 'shipping' THEN city END) AS shipping_city,
    MAX(CASE WHEN LOWER(address_type) = 'shipping' THEN state END) AS shipping_state,
    MAX(CASE WHEN LOWER(address_type) = 'shipping' THEN postcode END) AS shipping_postcode,
    MAX(CASE WHEN LOWER(address_type) = 'billing' THEN address_line_1 END) AS billing_address_line_1,
    MAX(CASE WHEN LOWER(address_type) = 'billing' THEN city END) AS billing_city,
    MAX(CASE WHEN LOWER(address_type) = 'billing' THEN state END) AS billing_state,
    MAX(CASE WHEN LOWER(address_type) = 'billing' THEN postcode END) AS billing_postcode
FROM gizmobox.bronze.v_addresses
GROUP BY customer_id;
```

### Write transformed data to the Silver

```sql
%sql
CREATE TABLE IF NOT EXISTS gizmobox.silver.addresses
AS
SELECT
    customer_id,
    MAX(CASE WHEN LOWER(address_type) = 'shipping' THEN address_line_1 END) AS shipping_address_line_1,
    MAX(CASE WHEN LOWER(address_type) = 'shipping' THEN city END) AS shipping_city,
    MAX(CASE WHEN LOWER(address_type) = 'shipping' THEN state END) AS shipping_state,
    MAX(CASE WHEN LOWER(address_type) = 'shipping' THEN postcode END) AS shipping_postcode,
    MAX(CASE WHEN LOWER(address_type) = 'billing' THEN address_line_1 END) AS billing_address_line_1,
    MAX(CASE WHEN LOWER(address_type) = 'billing' THEN city END) AS billing_city,
    MAX(CASE WHEN LOWER(address_type) = 'billing' THEN state END) AS billing_state,
    MAX(CASE WHEN LOWER(address_type) = 'billing' THEN postcode END) AS billing_postcode
FROM gizmobox.bronze.v_addresses
GROUP BY customer_id;
```

```sql
%sql
SELECT * FROM gizmobox.silver.addresses
```
