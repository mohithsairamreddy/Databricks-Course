### Extract Refunds Data using JDBC

```sql
%sql
DROP TABLE IF EXISTS gizmobox.bronze.refunds
```

```sql
CREATE CONNECTION IF NOT EXISTS gizmobox_azure_sql TYPE SQLSERVER
OPTIONS (
  host 'gizmobox-1-srv.database.windows.net',
  port '1433',
  user 'gizmoboxadm',
  password 'PASSWORD'
);
```

```sql
CREATE FOREIGN CATALOG IF NOT EXISTS gizmobox_azure_sql_catalog
USING CONNECTION gizmobox_azure_sql
OPTIONS (database 'gizmobox-db');
```

```sql
CREATE OR REPLACE TABLE gizmobox.bronze.refunds AS
SELECT * FROM gizmobox_azure_sql_catalog.dbo.refunds;
```
