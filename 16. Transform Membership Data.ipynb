## Transform Memberships Data
1. Extract customer_id from the file path
2. Write transformed data to the Silver

### Extract customer_id from the file path

```sql
%sql
SELECT * FROM gizmobox.bronze.v_memberships
```

```sql
%sql
SELECT regexp_extract(path, '.*/([^/]+)\\.png$', 1) AS customer_id,
content as membership_card
FROM gizmobox.bronze.v_memberships;
```

### Write transformed data to the Silver

```sql
%sql
CREATE TABLE IF NOT EXISTS gizmobox.silver.memberships
SELECT regexp_extract(path, '.*/([^/]+)\\.png$', 1) AS customer_id,
content as membership_card
FROM gizmobox.bronze.v_memberships;
```
