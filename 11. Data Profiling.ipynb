## Profile Data Manually

### COUNT

```python
%sql
SELECT COUNT(*),COUNT(customer_id),count(telephone),count(email) FROM gizmobox.bronze.v_customers
```

### COUNT_IF

```python
%sql
SELECT count_if(customer_id IS NULL),count_if(telephone IS NULL),count_if(email IS NULL) FROM gizmobox.bronze.v_customers
```

### WHERE

```python
%sql
SELECT * FROM gizmobox.bronze.v_customers
WHERE customer_id IS NULL
```

### DISTINCT

```python
%sql
SELECT COUNT(*) as record_count,COUNT(DISTINCT customer_id) as unique_record_count FROM gizmobox.bronze.v_customers
```
