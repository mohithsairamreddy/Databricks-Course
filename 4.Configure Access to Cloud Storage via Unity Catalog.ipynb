## CREATE EXTERNAL LOCATION

```python
%fs ls 'abfss://demo@decourseextdl2.dfs.core.windows.net/'
```

```python
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS dea_course_ext_dl_demo
url 'abfss://demo@decourseextdl2.dfs.core.windows.net/'
WITH(STORAGE CREDENTIAL de_course_ext_sc);
```

```python
%fs ls 'abfss://gizmobox@gizmobox.dfs.core.windows.net/landing/'
```

```python
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS dea_gizmobox_ext_dl
url 'abfss://gizmobox@gizmobox.dfs.core.windows.net/'
WITH(STORAGE CREDENTIAL gizmobox_sc);
```

## CREATE CATALOG

```python
%sql SHOW CATALOGS
```

```python
%sql
CREATE CATALOG IF NOT EXISTS  gizmobox
      MANAGED LOCATION 'abfss://gizmobox@gizmobox.dfs.core.windows.net/' 
      COMMENT 'GizmoBox Data Lake Catalog' 
```

```python
%sql SHOW CATALOGS
```

## CREATE SCHEMA

```python
%sql
USE CATALOG gizmobox;
    
CREATE  SCHEMA IF NOT EXISTS gizmobox.landing
    MANAGED LOCATION 'abfss://gizmobox@gizmobox.dfs.core.windows.net/landing';


CREATE  SCHEMA IF NOT EXISTS gizmobox.bronze
    MANAGED LOCATION 'abfss://gizmobox@gizmobox.dfs.core.windows.net/bronze';

CREATE  SCHEMA IF NOT EXISTS gizmobox.silver
    MANAGED LOCATION 'abfss://gizmobox@gizmobox.dfs.core.windows.net/silver';

CREATE  SCHEMA IF NOT EXISTS gizmobox.gold
    MANAGED LOCATION 'abfss://gizmobox@gizmobox.dfs.core.windows.net/gold';
```

```python
%sql SHOW SCHEMAS
```
